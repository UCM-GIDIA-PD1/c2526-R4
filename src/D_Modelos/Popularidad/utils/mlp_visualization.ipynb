```python
from src.utils.files import read_file
from src.utils.config import popularidad_mlp_file
from src.D_Modelos.Popularidad.mlp import MLPPopularity

minio_config = {"minio_write": False, "minio_read": False}

model_data = read_file(popularidad_mlp_file, minio_config)

modelo_sklearn = model_data["model"]
pipeline = modelo_sklearn.regressor_ 
keras_wrapper = pipeline.named_steps['mlp'] 
modelo_keras = keras_wrapper.model_

print("¡Modelo Keras extraído con éxito!")
```

```python
modelo_keras.summary()
```

```python
print("Estructura de pesos por capa")
for i, layer in enumerate(modelo_keras.layers):
    pesos_capa = layer.get_weights()
    # Si la capa tiene pesos (algunas como Dropout o Input no tienen)
    if pesos_capa: 
        # pesos_capa[0] son los Weights, pesos_capa[1] son los Biases
        print(f"Capa {i} ({layer.name}): Weights shape {pesos_capa[0].shape}, Biases shape {pesos_capa[1].shape}")
    else:
        print(f"Capa {i} ({layer.name}): Sin parámetros entrenables")

print("\nPesos reales de la primera capa oculta")
pesos_primera_capa = modelo_keras.layers[1].get_weights()[0] 
print(pesos_primera_capa)
```

```python
# Ver la configuración de compilación
config = modelo_keras.get_compile_config()

print("Optimizador:", config['optimizer'])
print("Función de pérdida:", config['loss'])
```
